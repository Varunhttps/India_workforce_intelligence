```python

%%writefile /content/drive/MyDrive/India_Workforce_Intelligence/agents/source_discovery.py

import os
import requests
import pandas as pd
from datetime import datetime, timezone
from urllib.parse import urlparse

TAVILY_URL = "https://api.tavily.com/search"

INDUSTRY_QUERIES = {
    "Automobile": [
        "India automobile industry workforce skill shortage labour shortage report",
        "India automotive manufacturing employee turnover productivity labour challenges"
    ],
    "Manufacturing": [
        "India manufacturing industry labour shortage skills gap workforce report",
        "India manufacturing absenteeism overtime productivity workforce challenges"
    ],
    "Food Processing": [
        "India food processing industry workforce labour shortage skill gaps report",
        "India food manufacturing employee availability productivity labour challenges"
    ]
}

def discover_sources(api_key, max_results=5):

    records = []

    headers = {
        "Content-Type": "application/json"
    }

    for industry, queries in INDUSTRY_QUERIES.items():

        for query in queries:

            print(f"\nSearching: {industry} | {query}")

            payload = {
                "api_key": api_key,
                "query": query,
                "topic": "general",
                "search_depth": "basic",
                "max_results": max_results,
                "include_answer": False,
                "include_raw_content": False
            }

            try:
                response = requests.post(
                    TAVILY_URL,
                    json=payload,
                    headers=headers,
                    timeout=60
                )

                response.raise_for_status()
                data = response.json()

                for result in data.get("results", []):

                    url = result.get("url", "")
                    domain = urlparse(url).netloc.lower()

                    records.append({
                        "industry": industry,
                        "search_query": query,
                        "title": result.get("title"),
                        "url": url,
                        "domain": domain,
                        "snippet": result.get("content"),
                        "search_score": result.get("score"),
                        "published_date": result.get("published_date"),
                        "collected_at": datetime.now(
                            timezone.utc
                        ).isoformat(),
                        "source_status": "Discovered - Not Yet Verified"
                    })

                print("Results found:", len(data.get("results", [])))

            except requests.exceptions.RequestException as error:
                print("Search request failed:", error)

    if not records:
        print("\nNo results collected.")
        return pd.DataFrame()

    df = pd.DataFrame(records)

    # Remove duplicate URLs
    df = df.drop_duplicates(subset=["url"])

    # Keep only valid HTTP/HTTPS links
    df = df[df["url"].str.startswith(("http://", "https://"))]

    return df.reset_index(drop=True)


def save_discovery_results(df, output_dir):

    os.makedirs(output_dir, exist_ok=True)

    timestamp = datetime.now(
        timezone.utc
    ).strftime("%Y%m%d_%H%M%S")

    csv_path = os.path.join(
        output_dir,
        f"discovered_sources_{timestamp}.csv"
    )

    json_path = os.path.join(
        output_dir,
        f"discovered_sources_{timestamp}.json"
    )

    df.to_csv(csv_path, index=False)
    df.to_json(json_path, orient="records", indent=4)

    print("\nSaved CSV:", csv_path)
    print("Saved JSON:", json_path)

    return csv_path, json_path
```

```python

%%writefile /content/drive/MyDrive/India_Workforce_Intelligence/agents/document_collector.py

import os
import re
import time
import hashlib
import requests
import pandas as pd
import trafilatura
import fitz

from bs4 import BeautifulSoup
from urllib.parse import urlparse
from datetime import datetime, timezone


HEADERS = {
    "User-Agent": "IndiaWorkforceIntelligenceResearch/1.0 (research prototype)"
}

MAX_FILE_SIZE = 15 * 1024 * 1024  # 15 MB
TIMEOUT = 30


def safe_filename(url):
    return hashlib.sha256(url.encode()).hexdigest()[:20]


def extract_pdf(content):
    """Extract text from a PDF held in memory."""
    document = fitz.open(stream=content, filetype="pdf")

    pages = []

    for page_number, page in enumerate(document, start=1):
        text = page.get_text("text")

        if text.strip():
            pages.append(
                f"\n--- PAGE {page_number} ---\n{text}"
            )

    document.close()
    return "\n".join(pages)


def extract_html(content):
    """Extract readable article content from HTML."""
    html = content.decode("utf-8", errors="replace")

    text = trafilatura.extract(
        html,
        include_comments=False,
        include_tables=True,
        favor_recall=True
    )

    if not text:
        soup = BeautifulSoup(html, "html.parser")

        for element in soup(
            ["script", "style", "nav", "footer", "header"]
        ):
            element.decompose()

        text = soup.get_text(separator="\n", strip=True)

    return text or ""


def collect_document(url):
    """Retrieve and extract one publicly accessible document."""

    result = {
        "url": url,
        "status": "failed",
        "content_type": None,
        "text": "",
        "error": None
    }

    try:
        parsed = urlparse(url)

        if parsed.scheme not in ("http", "https"):
            raise ValueError("Unsupported URL scheme")

        if not parsed.hostname:
            raise ValueError("Invalid URL")

        response = requests.get(
            url,
            headers=HEADERS,
            timeout=TIMEOUT,
            stream=True,
            allow_redirects=True
        )

        response.raise_for_status()

        # Do not accept redirects to non-HTTP destinations.
        if urlparse(response.url).scheme not in ("http", "https"):
            raise ValueError("Unsafe redirect destination")

        content_type = response.headers.get(
            "Content-Type", ""
        ).lower()

        result["content_type"] = content_type

        content = bytearray()

        for chunk in response.iter_content(chunk_size=65536):

            if not chunk:
                continue

            content.extend(chunk)

            if len(content) > MAX_FILE_SIZE:
                raise ValueError("Document exceeds 15 MB limit")

        content = bytes(content)

        if "pdf" in content_type or content.startswith(b"%PDF"):
            text = extract_pdf(content)

        elif "html" in content_type or "text/" in content_type:
            text = extract_html(content)

        else:
            raise ValueError(
                f"Unsupported content type: {content_type}"
            )

        if len(text.strip()) < 200:
            raise ValueError(
                "Insufficient readable text extracted"
            )

        result["text"] = text
        result["status"] = "success"

    except Exception as error:
        result["error"] = str(error)

    return result


def collect_documents(source_df, output_dir, max_documents=10):

    os.makedirs(output_dir, exist_ok=True)

    records = []

    # Limit the initial test batch
    batch = source_df.head(max_documents)

    for index, row in batch.iterrows():

        url = row["url"]

        print(f"\n[{len(records)+1}/{len(batch)}] Collecting: {url}")

        result = collect_document(url)

        record = {
            "industry": row.get("industry"),
            "title": row.get("title"),
            "url": url,
            "domain": row.get("domain"),
            "published_date": row.get("published_date"),
            "collection_status": result["status"],
            "content_type": result["content_type"],
            "text_length": len(result["text"]),
            "collected_at": datetime.now(
                timezone.utc
            ).isoformat(),
            "error": result["error"]
        }

        if result["status"] == "success":

            filename = safe_filename(url) + ".txt"

            filepath = os.path.join(output_dir, filename)

            with open(filepath, "w", encoding="utf-8") as file:
                file.write(result["text"])

            record["local_text_path"] = filepath

            print("Extracted characters:", len(result["text"]))

        else:
            record["local_text_path"] = None
            print("Collection failed:", result["error"])

        records.append(record)

        # Be respectful to source websites
        time.sleep(1)

    return pd.DataFrame(records)
```

```python

%%writefile /content/drive/MyDrive/India_Workforce_Intelligence/agents/problem_identifier.py

import os
import json
import re
import pandas as pd

from google import genai
from getpass import getpass

MODEL = "gemini-3.8-flash"

HR_CATEGORIES = [
    "Attendance Management",
    "Workforce Planning",
    "Labour Cost Management",
    "Productivity Management",
    "Skill Management",
    "Employee Performance",
    "Staffing and Workforce Availability",
    "HR Analytics and Reporting"
]

def create_client(api_key):
    return genai.Client(api_key=api_key)


def extract_problems(client, text, industry, title, url):

    # Limit each request to a manageable text segment.
    text = text[:12000]

    prompt = f"""
You are a rigorous workforce research analyst studying India.

DOCUMENT TITLE: {title}
INDUSTRY: {industry}
SOURCE URL: {url}

TASK:
Identify workforce management problems explicitly supported
by the supplied document text.

Allowed HR categories:
{HR_CATEGORIES}

RULES:
1. Only identify problems supported by the supplied text.
2. Never invent statistics, company practices or causes.
3. Do not convert recommendations into reported problems.
4. Separate explicit findings from reasonable interpretations.
5. If no relevant problem is found, return an empty list.
6. Evidence must be a short, verbatim passage from the text.
7. Do not infer company-specific attendance, shift or payroll
   practices from general industry statistics.
8. Do not recommend solutions.

Return valid JSON only, in this format:

{{
  "problems": [
    {{
      "problem_type": "Short description",
      "hr_category": "One allowed category",
      "description": "What the source reports",
      "evidence": "Exact supporting passage",
      "evidence_type": "Statistical / Employer Survey / Company Disclosure / News / Research Report / Other",
      "finding_type": "Explicit / Interpretation",
      "geography": "India or a supported Indian region",
      "confidence": 0.0
    }}
  ]
}}

DOCUMENT TEXT:
{text}
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt,
        config={
            "response_mime_type": "application/json",
            "temperature": 0.1
        }
    )

    try:
        result = json.loads(response.text)

        valid_problems = []

        for problem in result.get("problems", []):

            if problem.get("hr_category") not in HR_CATEGORIES:
                continue

            evidence = problem.get("evidence", "").strip()

            if not evidence or evidence not in text:
                continue

            problem["industry"] = industry
            problem["document_title"] = title
            problem["source_url"] = url

            valid_problems.append(problem)

        return valid_problems

    except (json.JSONDecodeError, TypeError):
        print("Could not parse model response for:", title)
        return []


def run_identification(metadata_file, output_file, api_key,
                       max_documents=2):

    client = create_client(api_key)

    documents = pd.read_csv(metadata_file)

    documents = documents[
        documents["collection_status"] == "success"
    ].head(max_documents)

    all_problems = []

    for _, document in documents.iterrows():

        path = document["local_text_path"]

        if not isinstance(path, str) or not os.path.exists(path):
            continue

        with open(path, "r", encoding="utf-8") as file:
            text = file.read()

        print("\nAnalysing:", document["title"])

        problems = extract_problems(
            client=client,
            text=text,
            industry=document["industry"],
            title=document["title"],
            url=document["url"]
        )

        print("Problems identified:", len(problems))

        all_problems.extend(problems)

    os.makedirs(os.path.dirname(output_file), exist_ok=True)

    with open(output_file, "w", encoding="utf-8") as file:
        json.dump(all_problems, file, indent=4, ensure_ascii=False)

    print("\nTotal identified problems:", len(all_problems))
    print("Saved:", output_file)

    return pd.DataFrame(all_problems)
```

```python

import os

BASE = "/content/drive/MyDrive/India_Workforce_Intelligence"
file_path = os.path.join(BASE, "agents", "problem_identifier.py")

with open(file_path, "r", encoding="utf-8") as f:
    code = f.read()

old = '''    response = client.models.generate_content(
        model=MODEL,
        contents=prompt,
        config={
            "response_mime_type": "application/json",
            "temperature": 0.1
        }
    )'''

new = '''    import time

    response = None

    for attempt in range(3):
        try:
            response = client.models.generate_content(
                model=MODEL,
                contents=prompt,
                config={
                    "response_mime_type": "application/json",
                    "temperature": 0.1
                }
            )
            break

        except Exception as error:
            message = str(error)

            if "503" in message or "UNAVAILABLE" in message:
                wait = 15 * (2 ** attempt)
                print(f"Gemini overloaded. Waiting {wait} seconds...")
                time.sleep(wait)
            else:
                raise

    if response is None:
        raise RuntimeError(
            "Gemini remained unavailable after 3 attempts. "
            "No findings were extracted; this is not a zero-problem result."
        )'''

if old not in code:
    print("Could not find the expected API block. No changes made.")
else:
    code = code.replace(old, new, 1)

    with open(file_path, "w", encoding="utf-8") as f:
        f.write(code)

    print("Retry mechanism successfully added!")
```

```python

import re
import unicodedata

def clean_for_comparison(text):
    text = unicodedata.normalize("NFKC", str(text))
    text = text.lower()

    # Join words split across PDF line breaks
    text = re.sub(r"-\s*\n\s*", "", text)

    # Ignore punctuation and formatting differences
    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()


# Use the full document, not the 12,000-character prompt excerpt
full_document = clean_for_comparison(full_text)

print("Full document characters:", len(full_text))
print("First 12,000 characters:", len(full_text[:12000]))

test_evidence = [
    "There are in-demand automotive jobs for which people are difficult to find",
    "There is a mismatch on quantity and quality between the demands of the industry",
    "information asymmetry between the demand and supply sides of the labour market"
]

for evidence in test_evidence:
    cleaned = clean_for_comparison(evidence)

    print("\nEvidence:", evidence)
    print("Found in full document:", cleaned in full_document)
    print("Found in first 12000:", cleaned in clean_for_comparison(full_text[:12000]))
```

```python

import pandas as pd
import json
from pathlib import Path

# Project folders
BASE = Path("/content/drive/MyDrive/India_Workforce_Intelligence")
DATA = BASE / "processed_data"

# Input files
findings_file = DATA / "identified_problems.json"
reports_file = DATA / "document_collection_results.csv"

# Output files
excel_file = DATA / "Workforce_Intelligence_Report.xlsx"
findings_csv = DATA / "Workforce_Findings.csv"

# 1. Load identified findings
with open(findings_file, "r", encoding="utf-8") as f:
    findings_data = json.load(f)

# Handle either a list or dictionary containing findings
if isinstance(findings_data, dict):
    findings_data = findings_data.get("findings", [])

findings_df = pd.json_normalize(findings_data)

# 2. Load collected report information
reports_df = pd.read_csv(reports_file)

# 3. Generate category summary
if not findings_df.empty:
    category_summary = (
        findings_df.groupby(["industry", "hr_category"])
        .size()
        .reset_index(name="Number_of_Findings")
    )
else:
    category_summary = pd.DataFrame(
        columns=["industry", "hr_category", "Number_of_Findings"]
    )

# 4. Export findings separately as CSV
findings_df.to_csv(findings_csv, index=False, encoding="utf-8-sig")

# 5. Create Excel workbook with three sheets
with pd.ExcelWriter(excel_file, engine="openpyxl") as writer:
    findings_df.to_excel(
        writer, sheet_name="Identified Findings", index=False
    )
    reports_df.to_excel(
        writer, sheet_name="Collected Reports", index=False
    )
    category_summary.to_excel(
        writer, sheet_name="Category Summary", index=False
    )

print("EXPORT COMPLETED!")
print("Excel:", excel_file)
print("CSV:", findings_csv)
print("Findings:", len(findings_df))
print("Collected reports:", len(reports_df))
```

```python

from pathlib import Path

file_path = Path(
    "/content/drive/MyDrive/India_Workforce_Intelligence/agents/source_discovery.py"
)

new_code = r'''
import os
import requests
import pandas as pd
from datetime import datetime, timezone
from urllib.parse import urlparse

TAVILY_URL = "https://api.tavily.com/search"

# Primary research focus
MANUFACTURING_QUERIES = {
    "Automotive": "global automotive manufacturing workforce challenges skilled labour shortages employee retention",
    "Electronics": "global electronics manufacturing workforce skills shortages labour challenges",
    "Textiles and Apparel": "global textile manufacturing workforce labour shortages productivity challenges",
    "Pharmaceuticals": "global pharmaceutical manufacturing workforce skills shortages labour challenges",
    "Food Processing": "global food processing manufacturing workforce challenges labour shortages",
    "Chemicals": "global chemical manufacturing workforce skills shortages labour challenges",
    "Heavy Machinery": "global heavy machinery manufacturing workforce shortages skills gaps",
    "Steel and Metals": "global steel metals manufacturing workforce labour shortages productivity challenges"
}

# Optional exploration beyond manufacturing
OTHER_INDUSTRY_QUERIES = {
    "Healthcare": "global healthcare workforce shortages staffing challenges",
    "IT and Technology": "global technology industry workforce skills shortages employment challenges",
    "Logistics": "global logistics warehousing workforce shortages labour challenges",
    "Retail": "global retail workforce employee shortages staffing challenges",
    "Banking and Finance": "global banking finance workforce skills shortages employment challenges",
    "Construction": "global construction workforce labour shortages skills gaps",
    "Hospitality": "global hospitality workforce staffing shortages employee turnover",
    "Energy": "global energy industry workforce skills shortages labour challenges",
    "Agriculture": "global agriculture workforce labour shortages employment challenges",
    "Education": "global education workforce teacher shortages staffing challenges"
}


def discover_sources(api_key, max_results=5, mode="manufacturing"):

    if mode == "manufacturing":
        selected_queries = MANUFACTURING_QUERIES
    elif mode == "other_industries":
        selected_queries = OTHER_INDUSTRY_QUERIES
    elif mode == "all":
        selected_queries = {
            **MANUFACTURING_QUERIES,
            **OTHER_INDUSTRY_QUERIES
        }
    else:
        raise ValueError(
            "Mode must be manufacturing, other_industries, or all"
        )

    records = []

    headers = {"Content-Type": "application/json"}

    for industry, query in selected_queries.items():

        print(f"\nSearching globally: {industry} | {query}")

        payload = {
            "api_key": api_key,
            "query": query,
            "topic": "general",
            "search_depth": "basic",
            "max_results": max_results,
            "include_answer": False,
            "include_raw_content": False
        }

        try:
            response = requests.post(
                TAVILY_URL,
                json=payload,
                headers=headers,
                timeout=60
            )

            response.raise_for_status()
            data = response.json()

            for result in data.get("results", []):

                url = result.get("url", "")
                if not url.startswith(("http://", "https://")):
                    continue

                records.append({
                    "geographic_scope": "Global",
                    "country": "Not specified",
                    "region": "Not specified",
                    "industry": industry,
                    "search_query": query,
                    "title": result.get("title"),
                    "url": url,
                    "domain": urlparse(url).netloc.lower(),
                    "snippet": result.get("content"),
                    "search_score": result.get("score"),
                    "published_date": result.get("published_date"),
                    "collected_at": datetime.now(
                        timezone.utc
                    ).isoformat(),
                    "source_status": "Discovered - Not Yet Verified"
                })

            print("Results found:", len(data.get("results", [])))

        except requests.exceptions.RequestException as error:
            print("Search request failed:", error)

    if not records:
        print("\nNo results collected.")
        return pd.DataFrame()

    df = pd.DataFrame(records)

    df = df.drop_duplicates(subset=["url"])
    return df.reset_index(drop=True)


def save_discovery_results(df, output_dir):

    os.makedirs(output_dir, exist_ok=True)

    timestamp = datetime.now(
        timezone.utc
    ).strftime("%Y%m%d_%H%M%S")

    csv_path = os.path.join(
        output_dir,
        f"discovered_sources_{timestamp}.csv"
    )

    json_path = os.path.join(
        output_dir,
        f"discovered_sources_{timestamp}.json"
    )

    df.to_csv(csv_path, index=False)
    df.to_json(json_path, orient="records", indent=4)

    print("\nSaved CSV:", csv_path)
    print("Saved JSON:", json_path)

    return csv_path, json_path
'''

file_path.write_text(new_code, encoding="utf-8")

print("Source Discovery Agent updated successfully!")
print("File:", file_path)
print("Size:", file_path.stat().st_size, "bytes")
```

```python

import pandas as pd
from urllib.parse import urlparse

def review_source(row):

    title = str(row["title"]).lower()
    snippet = str(row["snippet"]).lower()
    url = str(row["url"]).lower()

    text = title + " " + snippet
    domain = urlparse(url).netloc.lower()

    manufacturing_terms = [
        "manufacturing", "semiconductor", "factory",
        "production", "electronics", "industrial",
        "automotive", "textile", "pharmaceutical"
    ]

    workforce_terms = [
        "workforce", "worker", "workers", "labour",
        "labor", "skills", "talent", "shortage",
        "employment", "staffing", "employee",
        "skill gap", "workforce development"
    ]

    has_manufacturing = any(x in text for x in manufacturing_terms)
    has_workforce = any(x in text for x in workforce_terms)

    if "linkedin.com" in domain:
        return "Exclude - Restricted source"

    if "wikipedia.org" in domain:
        return "Review - Secondary source"

    if "youtube.com" in domain or "youtu.be" in domain:
        return "Review - Video source"

    if has_manufacturing and has_workforce:
        return "Potentially relevant - Verify document"

    if has_manufacturing or has_workforce:
        return "Manual review required"

    return "Likely irrelevant"


review_df = country_results.copy()

review_df["review_status"] = review_df.apply(
    review_source, axis=1
)

print("Source review summary:")
display(
    review_df["review_status"]
    .value_counts()
    .reset_index()
)

display(review_df)
```

```python

import os
import re
import time
import hashlib
import requests
import pandas as pd
import trafilatura
import fitz

from bs4 import BeautifulSoup
from urllib.parse import urlparse
from datetime import datetime, timezone


HEADERS = {
    "User-Agent": "IndiaWorkforceIntelligenceResearch/1.0 (research prototype)"
}

MAX_FILE_SIZE = 15 * 1024 * 1024  # 15 MB
TIMEOUT = 30


def safe_filename(url):
    return hashlib.sha256(url.encode()).hexdigest()[:20]


def extract_pdf(content):
    """Extract text from a PDF held in memory."""
    document = fitz.open(stream=content, filetype="pdf")

    pages = []

    for page_number, page in enumerate(document, start=1):
        text = page.get_text("text")

        if text.strip():
            pages.append(
                f"\n--- PAGE {page_number} ---\n{text}"
            )

    document.close()
    return "\n".join(pages)


def extract_html(content):
    """Extract readable article content from HTML."""
    html = content.decode("utf-8", errors="replace")

    text = trafilatura.extract(
        html,
        include_comments=False,
        include_tables=True,
        favor_recall=True
    )

    if not text:
        soup = BeautifulSoup(html, "html.parser")

        for element in soup(
            ["script", "style", "nav", "footer", "header"]
        ):
            element.decompose()

        text = soup.get_text(separator="\n", strip=True)

    return text or ""


def collect_document(url):
    """Retrieve and extract one publicly accessible document."""

    result = {
        "url": url,
        "status": "failed",
        "content_type": None,
        "text": "",
        "error": None
    }

    try:
        parsed = urlparse(url)

        if parsed.scheme not in ("http", "https"):
            raise ValueError("Unsupported URL scheme")

        if not parsed.hostname:
            raise ValueError("Invalid URL")

        response = requests.get(
            url,
            headers=HEADERS,
            timeout=TIMEOUT,
            stream=True,
            allow_redirects=True
        )

        response.raise_for_status()

        # Do not accept redirects to non-HTTP destinations.
        if urlparse(response.url).scheme not in ("http", "https"):
            raise ValueError("Unsafe redirect destination")

        content_type = response.headers.get(
            "Content-Type", ""
        ).lower()

        result["content_type"] = content_type

        content = bytearray()

        for chunk in response.iter_content(chunk_size=65536):

            if not chunk:
                continue

            content.extend(chunk)

            if len(content) > MAX_FILE_SIZE:
                raise ValueError("Document exceeds 15 MB limit")

        content = bytes(content)

        if "pdf" in content_type or content.startswith(b"%PDF"):
            text = extract_pdf(content)

        elif "html" in content_type or "text/" in content_type:
            text = extract_html(content)

        else:
            raise ValueError(
                f"Unsupported content type: {content_type}"
            )

        if len(text.strip()) < 200:
            raise ValueError(
                "Insufficient readable text extracted"
            )

        result["text"] = text
        result["status"] = "success"

    except Exception as error:
        result["error"] = str(error)

    return result


def collect_documents(source_df, output_dir, max_documents=10):

    os.makedirs(output_dir, exist_ok=True)

    records = []

    # Limit the initial test batch
    batch = source_df.head(max_documents)

    for index, row in batch.iterrows():

        url = row["url"]

        print(f"\n[{len(records)+1}/{len(batch)}] Collecting: {url}")

        result = collect_document(url)

        record = {
            "industry": row.get("industry"),
            "title": row.get("title"),
            "url": url,
            "domain": row.get("domain"),
            "published_date": row.get("published_date"),
            "collection_status": result["status"],
            "content_type": result["content_type"],
            "text_length": len(result["text"]),
            "collected_at": datetime.now(
                timezone.utc
            ).isoformat(),
            "error": result["error"]
        }

        if result["status"] == "success":

            filename = safe_filename(url) + ".txt"

            filepath = os.path.join(output_dir, filename)

            with open(filepath, "w", encoding="utf-8") as file:
                file.write(result["text"])

            record["local_text_path"] = filepath

            print("Extracted characters:", len(result["text"]))

        else:
            record["local_text_path"] = None
            print("Collection failed:", result["error"])

        records.append(record)

        # Be respectful to source websites
        time.sleep(1)

    return pd.DataFrame(records)
```

```python

%%writefile /content/drive/MyDrive/India_Workforce_Intelligence/agents/global_problem_identifier.py

import os
import json
import time
import pandas as pd

from google import genai

MODEL = "gemini-3.5-flash-lite"


def create_client(api_key):
    return genai.Client(api_key=api_key)


def extract_problems(client, text, metadata):

    # Keep a larger portion of the report for analysis.
    # We will improve this later using section-based chunking.
    text = text[:30000]

    prompt = f"""
You are a global workforce research analyst.

Identify explicitly reported workforce problems in this document.

Document title: {metadata.get('title')}
Industry: {metadata.get('industry')}
Search country: {metadata.get('country_searched')}
Known geographic scope: {metadata.get('actual_country')}

RULES:

1. Extract only actual workforce problems reported in the document.
2. Do not generate solutions or recommendations.
3. Do not invent statistics, companies, countries or facts.
4. Separate distinct workforce problems.
5. Distinguish skill shortages from skill gaps.
6. Identify staffing shortages, recruitment difficulties,
   retention problems, workforce planning issues and productivity
   problems only when explicitly supported.
7. Do not assume that the search country is the actual country.
8. For EU-wide findings, use "European Union" unless the document
   explicitly attributes the problem to an individual country.
9. If geography cannot be established, write "Not specified".
10. Do not invent company names.
11. Do not treat proposed solutions as workforce problems.
12. Return an empty list if no relevant problems are reported.
13. Do not include recommendations.

Return valid JSON in this format:

{{
  "problems": [
    {{
      "problem_type": "Short problem name",
      "description": "Description of the reported workforce problem",
      "country": "Country or geographic scope supported by the document",
      "region": "Geographic region supported by the document",
      "industry": "Industry",
      "sub_industry": "Specific manufacturing sub-industry",
      "company": null,
      "source_name": "Document title",
      "publication_date": null,
      "finding_type": "Explicit or Interpretation"
    }}
  ]
}}

Use null when information is unavailable.

DOCUMENT TEXT:
{text}
"""

    response = None

    for attempt in range(3):
        try:
            response = client.models.generate_content(
                model=MODEL,
                contents=prompt,
                config={
                    "response_mime_type": "application/json",
                    "temperature": 0.1
                }
            )
            break

        except Exception as error:
            message = str(error)

            if "503" in message or "UNAVAILABLE" in message:
                wait = 15 * (2 ** attempt)
                print(f"Gemini unavailable. Waiting {wait} seconds...")
                time.sleep(wait)
            else:
                raise

    if response is None:
        raise RuntimeError(
            "Gemini unavailable after three attempts."
        )

    result = json.loads(response.text)

    problems = []

    for problem in result.get("problems", []):

        problem["industry"] = metadata.get("industry")
        problem["source_name"] = metadata.get("title")
        problem["source_url"] = metadata.get("url")
        problem["publication_date"] = metadata.get(
            "published_date"
        )

        problems.append(problem)

    return problems


def run_identification(
    metadata_file,
    output_file,
    api_key,
    max_documents=3
):

    client = create_client(api_key)

    documents = pd.read_csv(metadata_file)

    # Only process successfully collected core manufacturing reports.
    documents = documents[
        (documents["collection_status"] == "success") &
        (documents["relevance_status"] == "Core manufacturing candidate")
    ].head(max_documents)

    all_problems = []

    for _, document in documents.iterrows():

        path = document["local_text_path"]

        if not isinstance(path, str) or not os.path.exists(path):
            print("Missing document:", document["title"])
            continue

        with open(path, "r", encoding="utf-8") as file:
            text = file.read()

        print("\nAnalysing:", document["title"])

        problems = extract_problems(
            client,
            text,
            document.to_dict()
        )

        print("Problems identified:", len(problems))

        all_problems.extend(problems)

    output_columns = [
        "problem_type",
        "description",
        "country",
        "region",
        "industry",
        "sub_industry",
        "company",
        "source_name",
        "source_url",
        "publication_date",
        "finding_type"
    ]

    df = pd.DataFrame(all_problems)

    if df.empty:
        df = pd.DataFrame(columns=output_columns)
    else:
        df = df.reindex(columns=output_columns)

    os.makedirs(os.path.dirname(output_file), exist_ok=True)

    df.to_json(
        output_file,
        orient="records",
        indent=4,
        force_ascii=False
    )

    print("\nTotal identified problems:", len(df))
    print("Saved:", output_file)

    return df
```

```python
start = global_code.find("def extract_problems(")
end = global_code.find("def run_identification(", start)

if start == -1 or end == -1:
    raise ValueError("Could not locate function boundaries.")

new_function = '''
def normalize_text(text):
    import re
    return re.sub(r"\\\\s+", " ", str(text)).strip()


def extract_problems(client, text, metadata):

    text = text[:30000]

    prompt = f"""
You are a global workforce research analyst.

Identify explicitly reported workforce problems in this document.

Document title: {metadata.get('title')}
Industry: {metadata.get('industry')}
Search country: {metadata.get('country_searched')}
Known geographic scope: {metadata.get('actual_country')}

RULES:
1. Extract only workforce problems explicitly supported by the document.
2. Do not generate solutions or recommendations.
3. Do not invent statistics, companies, countries or facts.
4. Avoid duplicate findings describing the same underlying problem.
5. Keep distinct occupational groups separate when supported by evidence.
6. Distinguish labour shortages, skill shortages, retention problems,
   recruitment difficulties and future talent gaps.
7. Use specific problem types, not generic repeated labels.
8. Do not assume search country is actual country.
9. For EU-wide findings, use country "European Union" and region "Europe".
10. Assign individual countries only when explicitly supported.
11. Do not infer geography from country mentions.
12. Use null when information is unavailable.
13. Return an empty list if no relevant problems are reported.
14. Use "Explicit" for directly supported findings.

Return JSON in this format:

{{
  "problems": [
    {{
      "problem_type": "Specific problem name",
      "description": "Description supported by the document",
      "country": null,
      "region": null,
      "industry": null,
      "sub_industry": null,
      "company": null,
      "source_name": null,
      "publication_date": null,
      "finding_type": "Explicit"
    }}
  ]
}}

DOCUMENT TEXT:
{text}
"""

    response = None

    for attempt in range(3):
        try:
            response = client.models.generate_content(
                model=MODEL,
                contents=prompt,
                config={
                    "response_mime_type": "application/json",
                    "temperature": 0.1
                }
            )
            break
        except Exception as error:
            message = str(error)
            if "503" in message or "UNAVAILABLE" in message:
                wait = 15 * (2 ** attempt)
                print(f"Gemini unavailable. Waiting {wait} seconds...")
                time.sleep(wait)
            else:
                raise

    if response is None:
        raise RuntimeError("Gemini unavailable after three attempts.")

    result = json.loads(response.text)
    problems = []
    seen = set()
    normalized_document = normalize_text(text)

    for problem in result.get("problems", []):

        description = normalize_text(problem.get("description", ""))

        if not description:
            continue

        # Require the description to be supported by the document.
        # The model's description may be a paraphrase, so we do not
        # reject it merely because it is not an exact quotation.

        evidence = normalize_text(problem.get("evidence", ""))

        if evidence and evidence not in normalized_document:
            print("Rejected unsupported evidence:", evidence[:100])
            continue

        country = problem.get("country")
        region = problem.get("region")

        if country == "European Union":
            region = "Europe"

        problem["country"] = country
        problem["region"] = region
        problem["industry"] = metadata.get("industry")
        problem["source_name"] = metadata.get("title")
        problem["source_url"] = metadata.get("url")
        problem["publication_date"] = metadata.get("published_date")
        problem["finding_type"] = "Explicit"

        # Exact duplicate protection within the same document.
        key = (
            str(problem.get("problem_type", "")).strip().lower(),
            description.lower(),
            str(country).strip().lower()
        )

        if key in seen:
            print("Skipped duplicate:", problem.get("problem_type"))
            continue

        seen.add(key)
        problems.append(problem)

    return problems

'''

global_code = global_code[:start] + new_function + global_code[end:]

with open(GLOBAL_FILE, "w", encoding="utf-8") as f:
    f.write(global_code)

print("Global extraction function updated and saved!")
```

```python
import re
from pathlib import Path

for _, row in metadata_df.iterrows():
    if row["collection_status"] != "success":
        continue

    path = Path(row["local_text_path"])

    print("\n" + "=" * 80)
    print("DOCUMENT:", row["title"])

    if not path.exists():
        print("Text file not found:", path)
        continue

    text = path.read_text(encoding="utf-8", errors="ignore")

    # Find likely publication-date references
    patterns = [
        r"(?i)publication date.{0,60}",
        r"(?i)published.{0,60}",
        r"(?i)date of publication.{0,60}",
        r"\b(?:January|February|March|April|May|June|July|August|September|October|November|December)\s+\d{4}\b",
        r"\b\d{1,2}\s+(?:January|February|March|April|May|June|July|August|September|October|November|December)\s+\d{4}\b",
        r"\b20\d{2}\b"
    ]

    matches = []
    for pattern in patterns:
        matches.extend(re.findall(pattern, text))

    print("Date references:", list(dict.fromkeys(matches))[:20])
    print("\nOpening text:")
    print(text[:1800])
```

```python
import os
import pandas as pd
from openpyxl import load_workbook
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from openpyxl.utils import get_column_letter
from openpyxl.worksheet.table import Table, TableStyleInfo
from openpyxl.chart import BarChart, Reference

# --------------------------------------------------
# 1. Prepare output folder and findings
# --------------------------------------------------

OUTPUT_DIR = os.path.join(BASE, "processed_data")
os.makedirs(OUTPUT_DIR, exist_ok=True)

excel_path = os.path.join(
    OUTPUT_DIR,
    "Global_Workforce_Intelligence_Report.xlsx"
)

findings = excel_findings.copy()

# Keep only useful manager-facing columns
finding_columns = [
    "problem_type",
    "description",
    "country",
    "region",
    "industry",
    "sub_industry",
    "company",
    "source_name",
    "source_url",
    "publication_date",
    "data_period",
    "review_status"
]

finding_columns = [
    col for col in finding_columns if col in findings.columns
]

findings = findings[finding_columns].copy()

# Replace missing values with blank cells in Excel
findings = findings.astype(object).where(
    pd.notna(findings), None
)

# --------------------------------------------------
# 2. Prepare source register
# --------------------------------------------------

sources = metadata_df.copy()

# Add review metadata where available
if "review_metadata" in globals():
    extra_columns = [
        col for col in [
            "title",
            "actual_country",
            "actual_region",
            "relevance_status",
            "verification_status",
            "review_notes",
            "countries_mentioned"
        ]
        if col in review_metadata.columns
    ]

    if "title" in extra_columns:
        sources = sources.merge(
            review_metadata[extra_columns].drop_duplicates("title"),
            on="title",
            how="left",
            suffixes=("", "_review")
        )

# Select available source fields
source_columns = [
    "title",
    "url",
    "industry",
    "country_searched",
    "actual_country",
    "actual_region",
    "published_date",
    "collection_status",
    "relevance_status",
    "verification_status",
    "review_notes"
]

source_columns = [
    col for col in source_columns if col in sources.columns
]

sources = sources[source_columns].copy()

# Rename for readability
sources = sources.rename(columns={
    "title": "Source title",
    "url": "Source URL",
    "industry": "Industry searched",
    "country_searched": "Country searched",
    "actual_country": "Actual geographic scope",
    "actual_region": "Region",
    "published_date": "Publication date",
    "collection_status": "Collection status",
    "relevance_status": "Relevance status",
    "verification_status": "Verification status",
    "review_notes": "Review notes"
})

sources = sources.astype(object).where(
    pd.notna(sources), None
)

# --------------------------------------------------
# 3. Prepare research summary
# --------------------------------------------------

summary_rows = [
    ["GLOBAL WORKFORCE INTELLIGENCE", ""],
    ["Report status", "Preliminary research – source-level verification pending"],
    ["Total findings", len(findings)],
    ["Documents collected", len(metadata_df)],
    ["Geographic scope", "Global"],
    ["Primary research focus", "Manufacturing"],
    ["Other industries", "Available as an exploration mode"],
    ["Evidence / Confidence / HR Category columns", "Excluded as requested"],
    ["Solutions and recommendations", "Not included"],
]

summary = pd.DataFrame(
    summary_rows,
    columns=["Metric", "Value"]
)

country_summary = (
    excel_findings["country"]
    .fillna("Not specified")
    .value_counts()
    .rename_axis("Country")
    .reset_index(name="Findings")
)

region_summary = (
    excel_findings["region"]
    .fillna("Not specified")
    .value_counts()
    .rename_axis("Region")
    .reset_index(name="Findings")
)

problem_summary = (
    excel_findings["problem_type"]
    .fillna("Not specified")
    .value_counts()
    .rename_axis("Problem type")
    .reset_index(name="Findings")
)

# --------------------------------------------------
# 4. Write all sheets
# --------------------------------------------------

with pd.ExcelWriter(excel_path, engine="openpyxl") as writer:

    findings.to_excel(
        writer, sheet_name="Workforce Findings", index=False
    )

    sources.to_excel(
        writer, sheet_name="Source Register", index=False
    )

    summary.to_excel(
        writer, sheet_name="Research Summary", index=False,
        startrow=0
    )

    country_summary.to_excel(
        writer, sheet_name="Research Summary", index=False,
        startrow=11
    )

    region_summary.to_excel(
        writer, sheet_name="Research Summary", index=False,
        startrow=11,
        startcol=3
    )

    problem_summary.to_excel(
        writer, sheet_name="Research Summary", index=False,
        startrow=11,
        startcol=6
    )

# --------------------------------------------------
# 5. Format workbook
# --------------------------------------------------

wb = load_workbook(excel_path)

header_fill = PatternFill(
    "solid", fgColor="1F4E78"
)

header_font = Font(
    color="FFFFFF", bold=True
)

title_fill = PatternFill(
    "solid", fgColor="D9EAF7"
)

thin_border = Border(
    bottom=Side(style="thin", color="D9E1F2")
)

for ws in wb.worksheets:

    ws.freeze_panes = "A2"
    ws.sheet_view.showGridLines = False

    # Format header rows
    for cell in ws[1]:
        cell.fill = header_fill
        cell.font = header_font
        cell.alignment = Alignment(
            horizontal="center",
            vertical="center",
            wrap_text=True
        )

    # Format cells and calculate column widths
    for column_cells in ws.columns:
        column_letter = get_column_letter(
            column_cells[0].column
        )

        max_length = 0

        for cell in column_cells:
            cell.alignment = Alignment(
                vertical="top",
                wrap_text=True
            )

            if cell.value is not None:
                max_length = max(
                    max_length,
                    min(len(str(cell.value)), 55)
                )

            cell.border = thin_border

        ws.column_dimensions[column_letter].width = min(
            max(max_length + 3, 15), 55
        )

    ws.row_dimensions[1].height = 30

# Add Excel filter tables
for sheet_name, table_name in [
    ("Workforce Findings", "FindingsTable"),
    ("Source Register", "SourcesTable")
]:
    ws = wb[sheet_name]

    if ws.max_row >= 2 and ws.max_column >= 1:
        ref = f"A1:{get_column_letter(ws.max_column)}{ws.max_row}"

        tab = Table(
            displayName=table_name,
            ref=ref
        )

        tab.tableStyleInfo = TableStyleInfo(
            name="TableStyleMedium2",
            showFirstColumn=False,
            showLastColumn=False,
            showRowStripes=True,
            showColumnStripes=False
        )

        ws.add_table(tab)

# Format summary sheet
ws = wb["Research Summary"]
ws.freeze_panes = "A2"

ws.merge_cells("A1:B1")
ws["A1"].fill = title_fill
ws["A1"].font = Font(
    size=16, bold=True, color="1F1F1F"
)

# Style the three summary tables
for header_row, start_col, end_col in [
    (12, 1, 2),
    (12, 4, 5),
    (12, 7, 8)
]:
    for col in range(start_col, end_col + 1):
        cell = ws.cell(header_row, col)
        cell.fill = header_fill
        cell.font = header_font
        cell.alignment = Alignment(
            horizontal="center",
            wrap_text=True
        )

ws.column_dimensions["A"].width = 34
ws.column_dimensions["B"].width = 65
ws.column_dimensions["D"].width = 25
ws.column_dimensions["E"].width = 15
ws.column_dimensions["G"].width = 38
ws.column_dimensions["H"].width = 15

# Add chart for findings by country
chart = BarChart()
chart.type = "bar"
chart.style = 10
chart.title = "Findings by Country"
chart.y_axis.title = "Country"
chart.x_axis.title = "Number of findings"

data = Reference(
    ws, min_col=2, min_row=12, max_row=11 + len(country_summary)
)
categories = Reference(
    ws, min_col=1, min_row=13, max_row=12 + len(country_summary)
)

chart.add_data(data, titles_from_data=True)
chart.set_categories(categories)
chart.height = 6
chart.width = 11

ws.add_chart(chart, "A25")

wb.save(excel_path)

# --------------------------------------------------
# 6. Confirm output
# --------------------------------------------------

print("EXCEL WORKBOOK CREATED SUCCESSFULLY")
print("File:", excel_path)
print("Findings:", len(findings))
print("Source register entries:", len(sources))
print("Sheets:", wb.sheetnames)
print("File size:", round(os.path.getsize(excel_path) / 1024, 2), "KB")
```

```python
from openpyxl import load_workbook
from collections import Counter

workbook_path = PROCESSED / "Global_Workforce_Intelligence_Report.xlsx"

wb = load_workbook(workbook_path)

# 1. Update CSET source register
ws = wb["Source Register"]

for row in range(2, ws.max_row + 1):
    title = str(ws.cell(row, 1).value or "")

    if "Strengthening the U.S. Semiconductor" in title:
        ws.cell(row, 7).value = "2026-09"
        ws.cell(row, 10).value = "Partially verified - page-level review"
        ws.cell(row, 11).value = (
            "Six findings checked against PDF pages 7, 11, 13, 13–14, 14 and 19. "
            "Construction labour shortage retained as an adjacent issue, outside the report's main scope."
        )
        break

# 2. Read findings
findings_ws = wb["Workforce Findings"]

headers = [cell.value for cell in findings_ws[1]]
records = []

for values in findings_ws.iter_rows(min_row=2, values_only=True):
    if any(value is not None for value in values):
        records.append(dict(zip(headers, values)))

country_counts = Counter(
    str(r["country"]) for r in records if r.get("country")
)
region_counts = Counter(
    str(r["region"]) for r in records if r.get("region")
)
problem_counts = Counter(
    str(r["problem_type"]) for r in records if r.get("problem_type")
)

# 3. Update summary
summary = wb["Research Summary"]

summary["B4"] = len(records)
summary["B5"] = "4 usable documents; 1 blocked by CAPTCHA"

# Clear old summary breakdowns
for row in range(13, summary.max_row + 1):
    for col in [1, 2, 4, 5, 7, 8]:
        summary.cell(row, col).value = None

# Rebuild country, region and problem type breakdowns
for i, (name, count) in enumerate(sorted(country_counts.items()), start=13):
    summary.cell(i, 1).value = name
    summary.cell(i, 2).value = count

for i, (name, count) in enumerate(sorted(region_counts.items()), start=13):
    summary.cell(i, 4).value = name
    summary.cell(i, 5).value = count

for i, (name, count) in enumerate(problem_counts.most_common(), start=13):
    summary.cell(i, 7).value = name
    summary.cell(i, 8).value = count

wb.save(workbook_path)

print("Workbook updated successfully.")
print("Total findings:", len(records))
print("Country breakdown:", dict(country_counts))
print("Region breakdown:", dict(region_counts))
print("Problem types:", dict(problem_counts))
```

```python
import pandas as pd

workbook_path = PROCESSED / "Global_Workforce_Intelligence_Report.xlsx"

df = pd.read_excel(workbook_path, sheet_name="Workforce Findings")

# Check exact duplicate rows
duplicates = df[df.duplicated(keep=False)]

# Check missing important fields
required_fields = [
    "problem_type", "description", "country",
    "region", "industry", "source_name", "source_url"
]

missing = df[df[required_fields].isna().any(axis=1)]

# Check source URL formatting
url_issues = df[
    df["source_url"].astype(str).str.startswith("[")
    | ~df["source_url"].astype(str).str.contains(
        r"^https?://", regex=True, na=False
    )
]

print("Total findings:", len(df))
print("Exact duplicate rows:", len(duplicates))
print("Rows with missing important fields:", len(missing))
print("Rows with URL formatting issues:", len(url_issues))

if not duplicates.empty:
    display(duplicates)

if not missing.empty:
    display(missing[required_fields])

if not url_issues.empty:
    display(url_issues[["source_name", "source_url"]])
```
